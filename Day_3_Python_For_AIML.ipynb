```python
# With Return
def add_and_square(a, b):
    sum_val = a + b
    return sum_val * sum_val

result = add_and_square(2, 3)
print(result)
```

```python
# With print
def add_and_square(a, b):
    sum_val = a + b
    print sum_val * sum_val

result = add_and_square(2, 3)
print(result)
```

return vs print:
- print is for display
- return is for logic and reuse
- ML code always depends on return

```python
##Function with loops
def sum_list(a):
  total=0
  for i in a:
    total=total+i
  return total
```

```python
a=[i for i in range(1,11)]
print(a)
sum_list(a)
```

```python
def retun_even(a):
  result=[]
  for i in a:
    if i%2==0:
      result.append(i)
  return result
```

```python
retun_even(a)
```

```python
b=[i for i in range(1,11,2)]
print(b)
retun_even(b)
```

```python
c=[]
retun_even(c)
```
