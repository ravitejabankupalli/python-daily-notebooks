Why reshape exists (intuition first)

Often you receive data as:

a flat list

a CSV column

a stream of numbers

But models expect:

rows = samples

columns = features

reshape enforces structure without changing data.

```python
a=np.array([1,2,3,4,5,6])
a
```

```python
a.shape
```

```python
x=a.reshape(2,3)
x
```

```python
#one of the golden rule of reshaping is all the elements in the original array and the reshaped array must be same and should have exact same count
#if we try to reshape it to a irregular shape then it throws an error
a=np.array([1,2,3,4,5,6])
print(a)
a.reshape(3,3)
```

Automatic dimension inference

it uses -1 as reshaping parameter like (2,-1) or (-1,3) where we are asking it to determine the shape accordingly

```python
a.reshape(2,-1)
```

```python
a.reshape(-1,2)
```

Reshape:
- Changes array structure, not data
- Total elements must match
- -1 lets NumPy infer dimensions
- Shape correctness is critical in ML


```python
import numpy as np

x = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
```

```python
x
```

```python
#we all know normal indexing like as shown below
print(x[0])
```

```python
print(x[1,2])
```

```python
##There is a way to access the entire column as shown below

print(x[0,:])

# What we are saying here is x[0th row, starting column: ending column]
# The actual way is x[Starting Row:Ending Row(Excluded), starting column: ending column(Excluded)]

#This is used for:

#feature selection

#normalization

#statistics
```

```python
x[0:2,0:3]
```

```python
## using this way we can modify the values as well
x[0,2]=1000
```

```python
x
```

Indexing:
- X[row] → one sample
- X[:, col] → one feature
- X[row, col] → single value
- Slicing enables feature/sample selection


```python
x
```

```python
## Lets discuss about axis
## Axis tells numpy to select a direction to apply a functiomn or to do anything
print(np.sum(x[1])) # it adds all the values in 2nd row of the x matrix
```

```python
## Now lets change the axis
print(np.sum(x,axis=0)) # NumPy goes down the rows, Sums each column, Result = one value per feature
```

```python
print(np.sum(x,axis=1))
```

```python
# One of the important usecases of "axis" is as shown
print("Row wise 1st elements mean: ",x.mean(axis=0))
print("Column wise 1st elements mean: ",x.mean(axis=1))
```

- axis = direction you collapse
- axis=0 → rows collapse → feature-wise result
- axis=1 → columns collapse → sample-wise result

Axis concept:
- rows = samples, columns = features
- axis=0 → operate column-wise
- axis=1 → operate row-wise
- Wrong axis leads to silent ML bugs
