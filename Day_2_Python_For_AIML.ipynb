```python
##while runs untill the condition is false
count=0
while count<=5:
  print(count)
  count+=1
```

```python
a=8
for i in range(0,10):
  if i==7:
    print("Break")
    break
  print(i)
```

```python
## BREAK statement stops the iteration, but continue skips the iteration as shown in the below example
a=5
for i in range(0,10):
  if i==5:
    print("Continue")
    continue
  print(i)
```
