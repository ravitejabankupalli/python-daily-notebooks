Aggregations:
- sum, mean, std summarize data
- axis controls feature-wise vs sample-wise aggregation
- mean and std are critical for normalization

### Now these are very important topics:

- **gradient descent**

- **stable training**

- **faster convergence**

```python
import numpy as np

x = np.array([
    [1, 200],
    [2, 300],
    [3, 400]
])
```

```python
mean=x.mean(axis=0)
standard_div=x.std(axis=0)
print("mean",mean)
print("standard_div",standard_div)
```

```python
x_norm=(x-mean)/standard_div
print(x_norm)
```

```python
#🧠 What just happened?

#Each feature now has:

#mean ≈ 0

#std ≈ 1

# This is standardization.
```

Normalization:
- rescales features to similar ranges
- improves training stability
- uses mean and standard deviation
- critical for gradient-based models


### Why do we normalize data?
**To center features around zero and scale them to comparable ranges so that optimization is stable and no feature dominates learning due to scale alone.**
